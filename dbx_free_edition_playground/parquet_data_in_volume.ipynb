```sql
%python
import pyspark.sql.functions
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType
```

```sql
%python
orders_schema = StructType([
    StructField('order_id', IntegerType()),
    StructField('order_date',TimestampType()),
    StructField('order_customer_id', IntegerType()),
    StructField('order_status', StringType())
])
df = spark.read.csv('/Volumes/workspace/lk_sandbox_db/lk_sandbox_volume/sample_datasets/orders_1gb.csv',schema = orders_schema)
```

```sql
%python
df1 = df.filter((df.order_id >= 201) & (df.order_id <= 400)).limit(200)
display(df1)
```

```sql
%python
df1.write.format('delta').mode('append').save('/Volumes/workspace/lk_sandbox_db/lk_sandbox_volume/delta_tables/orders')
```
